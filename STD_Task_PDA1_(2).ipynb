```python
"""
Step 1 – Load Dataset

Import pandas.
Read the dataset from the URL above into a DataFrame.
Show the first 10 rows.

Step 2 – Explore Data

Print the number of rows and columns in the dataset.
Show all column names.
Use .describe(include="all") to get summary statistics.
Step 3 – Analytics

How many Movies vs TV Shows are in the dataset?
Which are the top 5 countries producing Netflix content?
Which year had the most Netflix releases?


Step 4 – Visualisation

Create a bar chart comparing Movies vs TV Shows.
Create a line chart showing number of releases per year.

"""
```

```python
# Step 1 – Load Dataset

# Import pandas.
# Read the dataset from the URL above into a DataFrame.
# Show the first 10 rows.

import pandas as pd
import matplotlib.pyplot as plt

url = "/content/netflix_titles.csv"
df = pd.read_csv(url)

df.head()
```

```python
#Step 2 – Explore Data

#Print the number of rows and columns in the dataset.
#Show all column names.
# Use .describe(include="all") to get summary statistics.

print("\nShape of dataset:", df.shape)

print("\nColumn names:", df.columns.tolist())

print("\nSummary statistics:")
print(df.describe(include="all"))
```

```python
#Step 3 – Analytics

#How many Movies vs TV Shows are in the dataset?
#Which are the top 5 countries producing Netflix content?
# Which year had the most Netflix releases?

print("\nMovies vs TV Shows:")
print(df["type"].value_counts())

print("\nTop 5 countries:")
print(df["country"].value_counts().head(5))

print("\nReleases per year:")
print(df["release_year"].value_counts().head(1))

print("\nMost common rating:")
print(df["rating"].value_counts().head(1))
```

```python
# Step 4 – Visualisation

# Create a bar chart comparing Movies vs TV Shows.


df["type"].value_counts().plot(kind="bar", color=["skyblue", "lightgreen"], title="Movies vs TV Shows")
plt.xlabel("Type")
plt.ylabel("Count")
plt.show()


# Create a line chart showing number of releases per year.
df["release_year"].value_counts().sort_index().plot(kind="line", marker="o", title="Netflix Releases per Year")
plt.xlabel("Year")
plt.ylabel("Number of Releases")
plt.show()
```
